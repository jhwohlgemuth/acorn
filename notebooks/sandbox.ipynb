```rust
use data_encoding::BASE64;

let this = "foo";
let that = "bar";
let data = vec![this, that].join(";");
let encoded = BASE64.encode(data.as_bytes());
println!("{}", encoded);
let decoded = BASE64.decode(encoded.as_bytes()).unwrap();
println!("{:?}", str::from_utf8(&decoded).unwrap());
```
